```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import streamplot
```

```python
alpha0 = 1
alpha1 = .1
phi = 5
beta0 = 0.1
beta1 = 1
eps = 1
Kl = 1/5
delta = 0.1
p0 = 0.0001
p1 = 0.5
k = 0.5
L0 = 0.5

L_star = 0.0443
V_star = (delta) / (eps * (1 - (L_star / Kl)))

#graphing L as x axis, V as y axis
#x = np.linspace(0, 0.2, num=12)
x = np.linspace(0, 0.2, num=12)

y = np.linspace(0, 1, num=12)
L, V = np.meshgrid(x, y)

u = eps*(1 - (L/Kl))*V*L - delta*L
v = (p0 + (p1/(1 + np.exp(-k*(L - L0)))))*V - phi*V*L

streamplot(L, V, u, v)

plt.scatter(0, 0, color='r')
plt.scatter(L_star, V_star, color='r')

plt.xlabel("Law Enforcement")
plt.ylabel("Violent Protesters")
plt.title("Phase Portrait of Protests")
plt.show()
```
