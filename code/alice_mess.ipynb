```python
alpha0 = 1
alpha1 = .1
fi = 5
beta0 = 0.1
beta1 = 1
eps = 1
Kl = 1/5
delta = 0.1
p_0 = 0.0001
p_1 = 0.5
K = 30
l_0 = 0.5
```

```python
import numpy as np

def newton_solve(p_0, p_1, K, l_0, fi, x0=0.0, tol=1e-15, max_iter=10000):
    L = x0

    for _ in range(max_iter):
        f = fi*L + np.exp(-K*(L - l_0)) * (fi * L - p_0) - p_0 - p_1
        fp = fi + np.exp(-K*(L - l_0)) * (fi - fi * L * K + p_0 * K)  # derivative wrt x
        x_new = L - f/fp
        if abs(x_new - L) < tol:
            return x_new
        L = x_new
    raise RuntimeError("Did not converge")

L = newton_solve(p_0, p_1, K, l_0, fi, x0=0.2)
print(L)
```

V = delta / (eps * (1 - L / Kl))
print(V)

```python
# Elements of Jacobian
one_one = p_0 + p_1 / (1 - np.exp(-K * (L - l_0))) - fi * L
one_two = (p_1 * V * K * np.exp(-K * ((L - l_0)))) / (1 + np.exp(-K * ((L - l_0)))) **2 - fi * V
two_one = eps * L - eps * L**2 / Kl
two_two = eps * V - 2 * eps * V * L / Kl - delta
```

```python
print(one_one)
print(one_two)
print(two_one)
print(two_two)
```

```python
# Find eigenvalues of jacobian
from scipy import linalg as la

jac_at_equil = np.array([[one_one, one_two], [two_one, two_two]])

eigvals = la.eigvals(jac_at_equil)

print(eigvals)
```
